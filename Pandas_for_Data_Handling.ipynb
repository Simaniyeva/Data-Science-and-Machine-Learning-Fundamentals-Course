```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

geyser=sns.load_dataset('geyser')
print(geyser)
```

```python
geyser['mc_duration']=geyser['duration']-geyser['duration'].mean()
geyser['mc_waiting']=geyser['waiting']-geyser['waiting'].mean()
print(geyser)
```

```python
geyser['norm_duration']=(geyser['duration']-geyser['duration'].mean())/geyser['duration'].std()
geyser['norm_waiting']=(geyser['waiting']-geyser['waiting'].mean())/geyser['waiting'].std()
print(geyser)
```

```python
geyser['Z1_duration']=(geyser['duration']-geyser['duration'].min())/(geyser['duration'].max()-geyser['duration'].min())
geyser['Z1_waiting']=(geyser['waiting']-geyser['waiting'].min())/(geyser['waiting'].max()-geyser['waiting'].min())
print(geyser)
```

```python
fig,((orig,mc),(norm,Z1))=plt.subplots(2,2)
geyser.plot('duration','waiting',kind='scatter',ax=orig)
geyser.plot('mc_duration','mc_waiting',kind='scatter',ax=mc)
geyser.plot('norm_duration','norm_waiting',kind='scatter',ax=norm)
geyser.plot('Z1_duration','Z1_waiting',kind='scatter',ax=Z1)
plt.show()
```

# Concatenate Dataframes

```python
student_data1={"student_id":["S1","S2","S3","S4","S5"],"name":["Danniella Fenton","Ryder Storey","Bryce Jensen","Ed Bernal","Kwame Morin"],"marks":[200,210,190,222,199]}
students_1=pd.DataFrame(student_data1)

student_data2={"student_id":["S4","S5","S6","S7","S8"],"name":["Scarlette Fisher","Carla Williamson","Dante Morse","Kaiser William","Madeeha Preston"],"marks":[201,200,198,219,201]}
students_2=pd.DataFrame(student_data2)

student_data3={"active":['Yes','No','Yes','Yes'],"hobby":['Football','Voleyball',"Tennis","Football"]}
students_3=pd.DataFrame(student_data3)

print(students_1)
print()
print(students_2)
print()
print(students_3)
```

```python
students_4=pd.concat([students_1,students_2])
print(students_4)
```

```python
students_5=pd.concat([students_4,students_3],axis=1)
print(students_5)
```

```python
students_6=pd.concat([students_4.reset_index(drop=True),students_3],axis=1,join="inner")
print(students_6)
```

```python
students_7=pd.concat([students_1,students_4])
print(students_7)
```

```python
students_8=pd.concat([students_1,students_4],keys=["students_1","students_4"],names=['DataFrame','Students'])
print(students_8)
```

#Joining DataFrames

```python
student_data1={"student_id":["S1","S2","S3","S4"],"name":["Danniella Fenton","Ryder Storey","Bryce Jensen","Ed Bernal"],"marks":[200,210,190,222]}
students_1=pd.DataFrame(student_data1)

student_data2={"student_id":["S4","S5","S6","S7","S8"],"name":["Scarlette Fisher","Carla Williamson","Dante Morse","Kaiser William","Madeeha Preston"],"marks":[201,200,198,219,201]}
students_2=pd.DataFrame(student_data2)

student_data3={"active":['Yes','No','Yes','Yes'],"hobby":['Football','Voleyball',"Tennis","Football"]}
students_3=pd.DataFrame(student_data3)

contacts={"phone number":["0124541144","+991548526","554547788"],"email":["serena@gmail.com","biden01@gmail.com","mixail@mail.ru"]}
contacts=pd.DataFrame(contacts)
print(students_1)
print()
print(students_2)
print()
print(students_3)
print(contacts)
```

```python
students_4=students_1.join(contacts, on=None, how="left")
print(students_4)
```

```python
students_5=students_1.join(contacts, on=None, how="right")
print(students_5)
```

```python
students_6=students_1.join(contacts, on=None, how="inner")
print(students_6)
```

```python
students_7=students_1.join(contacts, on=None, how="outer")
print(students_7)
```

```python
students_8=students_1.join(contacts, on=None, how="cross")
print(students_8)
```

```python
students_8=students_1.join(students_3,on=None,how="inner")
print(students_8)
```

# Merging DataFrames

```python
import pandas as pd


customers = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Country": ["USA", "UK", "USA", "Germany"]
})


orders = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105],
    "CustomerID": [1, 2, 2, 5, 3],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Tablet"]
})


products = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Tablet", "Monitor"],
    "Price": [1200, 25, 45, 300, 200]
})

print("Customers:\n", customers)
print("Orders:\n", orders)
print("Products:\n", products)
```

```python
print(orders.merge(customers,how="inner",on=None,left_on="CustomerID",right_on="CustomerID"))
```

```python
print(orders.merge(products,how="inner",on=None,left_on="Product",right_on="Product"))
```

```python
print(orders.merge(customers,how="left",on=None,left_on="CustomerID",right_on="CustomerID"))
```

```python
print(orders.merge(customers,how="right",on=None,left_on="CustomerID",right_on="CustomerID"))
```

```python
print(orders.merge(customers,how="outer",on=None,left_on="CustomerID",right_on="CustomerID"))
```

# Transpose & Pivot Functions



```python
sales_data = pd.DataFrame({
    "Product": ["Laptop", "Tablet", "Smartphone"],
    "Q1": [150, 200, 300],
    "Q2": [180, 210, 330],
    "Q3": [170, 220, 350],
    "Q4": [160, 230, 370]
})
sales_long = pd.DataFrame({
    "Store": ["London", "London", "London", "Paris", "Paris", "Berlin", "Berlin", "Berlin"],
    "Month": ["Jan", "Feb", "Mar", "Jan", "Feb", "Jan", "Feb", "Mar"],
    "Product": ["Laptop", "Laptop", "Laptop", "Tablet", "Tablet", "Smartphone", "Smartphone", "Smartphone"],
    "Sales": [100, 120, 140, 80, 95, 200, 210, 250]
})
```

```python
sales_data
```

```python
sales_data=sales_data.transpose()
sales_data
```

```python
sales_long
```

```python
sales_pivot_1=sales_long.pivot(index="Product",columns="Month",values="Sales")
sales_pivot_1
```

```python
sales_pivot_2=sales_long.pivot_table(values="Sales",index="Product",columns='Month',aggfunc="sum")
sales_pivot_2
```

```python
store_count=sales_long.pivot_table(values="Store",index="Product",columns='Month',aggfunc="nunique")
store_count
```
